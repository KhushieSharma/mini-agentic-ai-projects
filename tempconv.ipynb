```python
from langgraph.graph import StateGraph, START, END
from typing import TypedDict
```

```python
class Temperature(TypedDict):
    celsius: float
    fahrenheit: float



#conv
def convert(state:Temperature)->Temperature:
  celsius=state['celsius']
  fahrenheit=(celsius*9/5)+32
  state['fahrenheit']=round(fahrenheit,2)
  return state
```

```python
#graph
graph=StateGraph(Temperature)
graph.add_node('convert', convert)
graph.add_edge(START, 'convert')
graph.add_edge('convert', END)
wf=graph.compile()

init={'celsius': 100.0}
final=wf.invoke(init)
print(final)
```
